## 🎯 Objectives for Day 6:
1. ✅ Compute 95% historical Value at Risk (VaR) and CVaR for mutual fund schemes.
2. ✅ Calculate the Herfindahl-Hirschman Index (HHI) for each fund's portfolio weights to detect concentration risk.
3. ✅ Store and export advanced analytics metrics.
4. ✅ Plot risk and concentration profiles.

```python
import os
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt
import seaborn as sns

BASE_DIR = os.path.abspath(os.path.join(os.getcwd(), '..'))
DB_PATH = os.path.join(BASE_DIR, 'data', 'bluestock_mf.db')
CHARTS_DIR = os.path.join(BASE_DIR, 'outputs', 'charts')

conn = sqlite3.connect(DB_PATH)
df_advanced = pd.read_sql("SELECT * FROM fact_advanced_metrics", conn)
conn.close()

print("✅ Advanced Analytics loaded successfully.")
df_advanced.head(5)
```

## 📉 Value at Risk (VaR) Distribution Across Schemes

```python
plt.figure(figsize=(10, 5))
sns.histplot(df_advanced['daily_var_95_pct'], bins=12, kde=True, color='red')
plt.title("Distribution of 95% Daily Value at Risk (VaR) Across Funds", fontsize=14, fontweight='bold')
plt.xlabel("Daily VaR (% Loss)")
plt.ylabel("Scheme Count")
plt.tight_layout()
plt.savefig(os.path.join(CHARTS_DIR, "var_distribution.png"))
plt.show()
```

## 🏆 Portfolio Concentration Risk (HHI Index Profile)

```python
df_hhi = df_advanced.dropna(subset=['hhi_index'])
plt.figure(figsize=(10, 5))
sns.countplot(data=df_hhi, x='concentration_risk', palette='Set2')
plt.title("Portfolio Concentration Risk Profile (HHI Classifications)", fontsize=14, fontweight='bold')
plt.xlabel("Concentration Level Category")
plt.ylabel("Scheme Count")
plt.tight_layout()
plt.savefig(os.path.join(CHARTS_DIR, "hhi_concentration_distribution.png"))
plt.show()
```
