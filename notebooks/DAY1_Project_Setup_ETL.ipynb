## 📦 STEP 1 — Import Required Libraries

Initializing the environment by importing the standard data analysis libraries.

```python
import pandas as pd
import os
import warnings
warnings.filterwarnings('ignore')

# Configuring Pandas display settings for optimal viewing
pd.set_option('display.max_columns', 50)
pd.set_option('display.max_rows', 20)
pd.set_option('display.float_format', '{:.2f}'.format)

print('✅ Libraries successfully imported.')
print(f'📦 Pandas version: {pd.__version__}')
```

## 📂 STEP 2 — Define Directory Paths

Setting up absolute paths for the raw and processed data directories to ensure reproducibility.

```python
# Define project root and data directories
BASE_DIR = os.path.abspath(os.path.join(os.getcwd(), '..'))
RAW_DATA_DIR = os.path.join(BASE_DIR, 'data', 'raw')
PROCESSED_DIR = os.path.join(BASE_DIR, 'data', 'processed')

print(f'📁 Raw Data Directory: {RAW_DATA_DIR}\n')

# Verify all files exist in the directory
files = os.listdir(RAW_DATA_DIR)
print(f'📊 Total datasets detected: {len(files)}\n')
for f in sorted(files):
    print(f'  ✅ {f}')
```

## 📥 STEP 3 — Data Extraction

Loading all raw CSV files into Pandas DataFrames for initial inspection.

```python
# Load datasets
df_fund_master         = pd.read_csv(os.path.join(RAW_DATA_DIR, '01_fund_master.csv'))
df_nav_history         = pd.read_csv(os.path.join(RAW_DATA_DIR, '02_nav_history.csv'))
df_aum_fund_house      = pd.read_csv(os.path.join(RAW_DATA_DIR, '03_aum_by_fund_house.csv'))
df_monthly_sip         = pd.read_csv(os.path.join(RAW_DATA_DIR, '04_monthly_sip_inflows.csv'))
df_category_inflows    = pd.read_csv(os.path.join(RAW_DATA_DIR, '05_category_inflows.csv'))
df_folio_count         = pd.read_csv(os.path.join(RAW_DATA_DIR, '06_industry_folio_count.csv'))
df_scheme_performance  = pd.read_csv(os.path.join(RAW_DATA_DIR, '07_scheme_performance.csv'))
df_transactions        = pd.read_csv(os.path.join(RAW_DATA_DIR, '08_investor_transactions.csv'))
df_portfolio_holdings  = pd.read_csv(os.path.join(RAW_DATA_DIR, '09_portfolio_holdings.csv'))
df_benchmark           = pd.read_csv(os.path.join(RAW_DATA_DIR, '10_benchmark_indices.csv'))

print('✅ Data ingestion complete. All 10 datasets loaded into memory.')
```

## 📊 STEP 5 — Data Quality & Integrity Summary

Generating a consolidated report detailing dimensions and missing values across all tables.

```python
# Constructing a summary dataframe
datasets = {
    '01_fund_master':          df_fund_master,
    '02_nav_history':          df_nav_history,
    '03_aum_by_fund_house':    df_aum_fund_house,
    '04_monthly_sip_inflows':  df_monthly_sip,
    '05_category_inflows':     df_category_inflows,
    '06_industry_folio_count': df_folio_count,
    '07_scheme_performance':   df_scheme_performance,
    '08_investor_transactions':df_transactions,
    '09_portfolio_holdings':   df_portfolio_holdings,
    '10_benchmark_indices':    df_benchmark,
}

summary_data = []
for name, df in datasets.items():
    missing = df.isnull().sum().sum()
    summary_data.append({
        'Dataset': name,
        'Rows': df.shape[0],
        'Columns': df.shape[1],
        'Total Missing Values': missing
    })

df_summary = pd.DataFrame(summary_data)
print('='*60)
print('📊 MASTER DATASET SUMMARY')
print('='*60)
df_summary
```

## 💾 STEP 6 — Export Summary Log

Saving the generated ETL summary log to the processed data directory.

```python
# Export to processed directory
summary_path = os.path.join(PROCESSED_DIR, 'day1_dataset_summary.csv')
df_summary.to_csv(summary_path, index=False)

print(f'✅ Pipeline complete. Log saved to: {summary_path}')
```
