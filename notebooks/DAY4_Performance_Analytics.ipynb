## 🎯 Objectives for Day 4:
1. ✅ Retrieve mutual fund metrics and clean schemas.
2. ✅ Implement ranking logic using trailing returns, Sharpe, and expense ratios.
3. ✅ Build the weighted composite scorecard and identify the top 5 recommended schemes.
4. ✅ Compare performance profiles.

```python
import os
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt
import seaborn as sns

BASE_DIR = os.path.abspath(os.path.join(os.getcwd(), '..'))
DB_PATH = os.path.join(BASE_DIR, 'data', 'bluestock_mf.db')
CHARTS_DIR = os.path.join(BASE_DIR, 'outputs', 'charts')

conn = sqlite3.connect(DB_PATH)
df_scorecard = pd.read_sql("SELECT * FROM fact_scorecard", conn)
conn.close()

print("✅ Scorecard loaded successfully.")
df_scorecard.head(5)[['amfi_code', 'scheme_name', 'final_rank', 'composite_score']]
```

## 📈 Visualizing the Scorecard Distribution

```python
plt.figure(figsize=(10, 5))
sns.histplot(df_scorecard['composite_score'], bins=12, kde=True, color='purple')
plt.title("Distribution of Mutual Fund Composite Scores", fontsize=14, fontweight='bold')
plt.xlabel("Weighted Score")
plt.ylabel("Scheme Count")
plt.tight_layout()
plt.savefig(os.path.join(CHARTS_DIR, "composite_score_distribution.png"))
plt.show()
```

## 🏆 Top 5 Schemes Returns Comparison

```python
top_5 = df_scorecard.head(5)

plt.figure(figsize=(12, 6))
x = range(len(top_5))
width = 0.25

plt.bar([i - width for i in x], top_5['return_1yr_pct'], width, label='1Yr Return %', color='skyblue')
plt.bar(x, top_5['return_3yr_pct'], width, label='3Yr Return %', color='lightgreen')
plt.bar([i + width for i in x], top_5['return_5yr_pct'], width, label='5Yr Return %', color='salmon')

plt.xticks(x, top_5['scheme_name'], rotation=30, ha='right')
plt.title("Trailing Performance of Top 5 Recommended Schemes", fontsize=14, fontweight='bold')
plt.ylabel("Returns in %")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(CHARTS_DIR, "top_5_returns_comparison.png"))
plt.show()
```
